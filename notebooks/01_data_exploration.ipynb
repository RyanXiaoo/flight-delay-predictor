```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Adjust display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("Libraries imported.")
```

```python
# Define the path to the data file
# Check if the file is in the root or in the data directory
# Assuming notebook is in 'notebooks/' dir
data_file_root = '../flights_sample_3m.csv'
data_file_data_dir = '../data/flights_sample_3m.csv'

if os.path.exists(data_file_root):
    data_path = data_file_root
elif os.path.exists(data_file_data_dir):
    data_path = data_file_data_dir
else:
    # Try checking relative to current dir (if running notebook from root)
    data_file_root_alt = 'flights_sample_3m.csv'
    data_file_data_dir_alt = 'data/flights_sample_3m.csv'
    if os.path.exists(data_file_root_alt):
        data_path = data_file_root_alt
    elif os.path.exists(data_file_data_dir_alt):
        data_path = data_file_data_dir_alt
    else:
        raise FileNotFoundError("Could not find flights_sample_3m.csv in root or data/ directory.")

print(f"Loading data from: {data_path}")

# Load the dataset (consider using nrows for initial exploration if memory is an issue)
# df = pd.read_csv(data_path, nrows=100000) # Use this for quicker initial exploration
df = pd.read_csv(data_path)

print("Dataset loaded successfully.")
print(f"Shape: {df.shape}")
```

```python
# Initial inspection
print("First 5 rows:")
display(df.head())

print("\nBasic Info:")
# Use verbose=True to see all columns if needed, False for summary
df.info(verbose=False) 

print("\nDescriptive Statistics (Numerical Columns):")
display(df.describe())

print("\nDescriptive Statistics (Object Columns):")
display(df.describe(include='object'))
```

```python
# Check current data types
print("Initial data types:")
print(df.dtypes.value_counts())
print("\\nSample of object types:")
print(df.select_dtypes(include='object').head())

# Convert FL_DATE to datetime objects
print("\\nConverting FL_DATE to datetime...")
df['FL_DATE'] = pd.to_datetime(df['FL_DATE'])
print("FL_DATE converted.")

# We might need to handle time columns (like CRS_DEP_TIME) later.
# They are often stored as HHMM (e.g., 1430 for 2:30 PM).
# Let's check their format first.
print("\\nSample time columns:")
print(df[['CRS_DEP_TIME', 'DEP_TIME', 'CRS_ARR_TIME', 'ARR_TIME']].head())
print("\\nData types of time columns:")
print(df[['CRS_DEP_TIME', 'DEP_TIME', 'CRS_ARR_TIME', 'ARR_TIME']].dtypes)

# Check data types again after conversion
print("\\nUpdated data types (first 15):")
print(df.dtypes.head(15))
```

```python
# Check for missing values
print("Checking for missing values...")
missing_values = df.isnull().sum()
missing_percent = (missing_values / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': missing_percent
})

# Display columns with missing values, sorted by percentage
print("Columns with missing values (sorted):")
display(missing_df[missing_df['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False))

# Visualize missingness (optional, can be slow on large datasets)
# import missingno as msno
# print("\\nVisualizing missing values matrix (sample):")
# msno.matrix(df.sample(1000)) # Use a sample to avoid performance issues
# plt.show()

# print("\\nVisualizing missing values bar chart:")
# msno.bar(df)
# plt.show()
```
